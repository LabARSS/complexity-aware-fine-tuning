```python
import pandas as pd

model = "phi"
split = "valid"

easy_df = pd.read_csv(f"../../data/data_splits/entropy_fallback/{model}/{split}_df_easy.tsv", sep="\t")
mid_df = pd.read_csv(f"../../data/data_splits/entropy_fallback/{model}/{split}_df_middle.tsv", sep="\t")
hard_df = pd.read_csv(f"../../data/data_splits/entropy_fallback/{model}/{split}_df_hard.tsv", sep="\t")
```

```python
stacked_df = pd.concat([easy_df, mid_df, hard_df], ignore_index=True)
```

```python
stacked_df = stacked_df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
stacked_df.to_csv(f"../../data/data_splits/entropy_fallback/{model}/{split}_df_combined.tsv", sep="\t", index=False)
```
