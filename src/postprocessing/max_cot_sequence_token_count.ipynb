```python
import pandas as pd

df = pd.read_json("../../data/out/distillation/gsm8k_distilled_cot.jsonl", lines=True)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.2-3B-Instruct")

# Tokenize all distill_response values and find max token count
token_counts = df["distill_response"].apply(lambda x: len(tokenizer.encode(str(x))))
max_token_count = token_counts.max()
max_idx = token_counts.idxmax()

print(f"Maximum token count: {max_token_count}")
print(f"Row index with max tokens: {max_idx}")
```

```python
import sys
sys.path.append("../..")

from reasoning_fine_tune.prompts.gsm8k_cot_answer import cot_answer_prompt_from_row

# Load train_combined.tsv and find max token count for full prompt
df_train = pd.read_json("../../data/source/gsm8k/gsm8k_train.jsonl", lines=True)

prompt_token_counts = df_train.apply(lambda row: len(tokenizer.encode(cot_answer_prompt_from_row(row))), axis=1)
max_prompt_tokens = prompt_token_counts.max()
max_prompt_idx = prompt_token_counts.idxmax()

print(f"Maximum token count for prompt: {max_prompt_tokens}")
print(f"Row index with max tokens: {max_prompt_idx}")
```
